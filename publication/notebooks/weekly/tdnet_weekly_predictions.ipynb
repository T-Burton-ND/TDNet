# TDNet weekly predictions — blog-ready

Produces one Top-25-games figure with team logos and predicted margins, plus the complete weekly game table. Canonical CSV/Parquet tables, PNG/SVG figures, captions, alt text, and blog Markdown are written by shared package code.

```python
from pathlib import Path
import shutil
import sys
from IPython.display import Image, display, Markdown

PROJECT_ROOT = next(p for p in [Path.cwd().resolve(), *Path.cwd().resolve().parents] if (p / 'src' / 'gridiron_ml').exists())
SRC = PROJECT_ROOT / 'src'
if str(SRC) not in sys.path:
    sys.path.insert(0, str(SRC))

from gridiron_ml.publication import build_weekly_blog_package
```

## Weekly parameters
Update `WEEK` and the frozen inventory/poll snapshot before publication.

```python
SEASON = 2026
WEEK = 1
FINGERPRINT_VERSION = 0
MODEL_INVENTORY = PROJECT_ROOT / 'docs/publication_2026/weekly_learned_model_inventory.csv'
SCHEDULE_SNAPSHOT = PROJECT_ROOT / f'data/raw/cfbd/v2/games/{SEASON}.parquet'
# The official 2026 preseason AP poll is not yet in the repository. This
# illustrative default uses the latest frozen 2025 TDNet poll and labels it.
AP_TOP25_SNAPSHOT = PROJECT_ROOT / f'data/raw/cfbd/v2/rankings/{SEASON}.parquet'
TDNET_TOP25_SNAPSHOT = None  # Optional comparison; never drives the AP game slate.
TOP25_LABEL = 'AP Top 25'
OUTPUT_ROOT = PROJECT_ROOT / 'data' / 'publication' / str(SEASON) / 'weekly_predictions' / f'week_{WEEK:02d}' / 'blog_preview'
LOGO_DIR = PROJECT_ROOT / 'data/meta/logos/by_team'
```

```python
report = build_weekly_blog_package(project_root=PROJECT_ROOT, season=SEASON, week=WEEK, model_inventory_path=MODEL_INVENTORY, schedule_snapshot_path=SCHEDULE_SNAPSHOT, ap_top25_path=AP_TOP25_SNAPSHOT, tdnet_top25_path=TDNET_TOP25_SNAPSHOT, top25_label=TOP25_LABEL, fingerprint_version=FINGERPRINT_VERSION, output_root=OUTPUT_ROOT, logo_dir=LOGO_DIR)
FIGURE_ROOT = PROJECT_ROOT / 'publication' / str(SEASON) / 'figures' / 'weekly_predictions' / f'week_{WEEK:02d}'
FIGURE_ROOT.mkdir(parents=True, exist_ok=True)
for figure_path in report['figures'].values():
    shutil.copy2(figure_path, FIGURE_ROOT / Path(figure_path).name)
print(report['output_root'])
display(report['manifest'])
```

## AP Top 25 teams playing this week

```python
display(Image(filename=str(report['figures']['top25_matchups'])))
display(report['top25_games'])
```

```python
display(report['all_games'])
display(Image(filename=str(report['figures']['all_games_table'])))
```
